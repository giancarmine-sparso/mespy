# Basic workflow

Load a CSV with pandas, then create a histogram with `mespy.histogram` and a linear fit with `mespy.lin_fit`. Find the PDF plots and TOML fit results in your chosen output folder.

(bw-data)=
## Load your data

Save {download}`measurements.csv <data/measurements.csv>` in your project's `data/` folder. It has three columns: `x`, `y`, and `sigma_y`, the uncertainty on each `y` value.

```{literalinclude} data/measurements.csv
:language: text
```

Run the code from your project folder. pandas reads the CSV into a DataFrame named `df`:

````python
import pandas as pd
from mespy import histogram, lin_fit

df = pd.read_csv("data/measurements.csv")
````

Access any column with `df["column_name"]`. For example, `df["y"]` contains the measured values.

(bw-histogram)=
## Create a histogram

`histogram` saves a PDF at `save_path`. The `results/` folder must already exist: `histogram` does not create missing folders, so create it before saving.

````python
from pathlib import Path

Path("results").mkdir(exist_ok=True)

histogram(
    df["y"],
    save_path="results/histogram.pdf",
)
````

(bw-fit)=
## Perform a linear fit

Use the same DataFrame. Pass `sigma_y` to include the measurement uncertainties in the fit.

````python
lin_fit(
    df["x"],
    df["y"],
    sigma_y=df["sigma_y"],
    output="results/linear_fit",
)
````

`output` sets the destination and base filename, **without an extension**. This call creates:

- `linear_fit.pdf` — the fit plot.
- `linear_fit.toml` — the numerical fit results.

(bw-files)=
## Saved files

Your project now contains:

```text
project/
├── data/
│   └── measurements.csv
└── results/
    ├── histogram.pdf
    ├── linear_fit.pdf
    └── linear_fit.toml
```

**Both functions save PDF plots.** `lin_fit` also saves numerical results in TOML, a text file you can open in any text editor.
