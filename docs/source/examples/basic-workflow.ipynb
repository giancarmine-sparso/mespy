# Flusso di lavoro di base

Carica un CSV con pandas, poi crea un istogramma con `mespy.histogram` e un fit lineare con `mespy.lin_fit`. Trovi i grafici PDF e i risultati del fit in TOML nella cartella di destinazione scelta.

(bw-data)=
## Carica i tuoi dati

Salva {download}`measurements.csv <data/measurements.csv>` nella cartella `data/` del progetto. Contiene tre colonne: `x`, `y` e `sigma_y`, l'incertezza su ogni valore di `y`.

```{literalinclude} data/measurements.csv
:language: text
```

Esegui il codice dalla cartella del progetto. pandas legge il CSV in un DataFrame chiamato `df`:

````python
import pandas as pd
from mespy import histogram, lin_fit

df = pd.read_csv("data/measurements.csv")
````

Accedi a qualsiasi colonna con `df["column_name"]`. Per esempio, `df["y"]` contiene i valori misurati.

(bw-histogram)=
## Crea un istogramma

`histogram` salva un PDF in `save_path`. La cartella `results/` deve già esistere: `histogram` non crea le cartelle mancanti, quindi creala prima di salvare.

````python
from pathlib import Path

Path("results").mkdir(exist_ok=True)

histogram(
    df["y"],
    save_path="results/histogram.pdf",
)
````

(bw-fit)=
## Esegui un fit lineare

Usa lo stesso DataFrame. Passa `sigma_y` per includere le incertezze di misura nel fit.

````python
lin_fit(
    df["x"],
    df["y"],
    sigma_y=df["sigma_y"],
    output="results/linear_fit",
)
````

`output` imposta la destinazione e il nome base dei file, **senza estensione**. Questa chiamata crea:

- `linear_fit.pdf` — il grafico del fit.
- `linear_fit.toml` — i risultati numerici del fit.

(bw-files)=
## File salvati

Il progetto ora contiene:

```text
project/
├── data/
│   └── measurements.csv
└── results/
    ├── histogram.pdf
    ├── linear_fit.pdf
    └── linear_fit.toml
```

**Entrambe le funzioni salvano grafici PDF.** `lin_fit` salva anche i risultati numerici in TOML, un file di testo che puoi aprire con qualsiasi editor di testo.
